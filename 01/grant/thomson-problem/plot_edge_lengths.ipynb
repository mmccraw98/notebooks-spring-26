```python
import os
import numpy as np
import matplotlib.pyplot as plt

root = 'data'

for p in os.listdir(root):
    data = np.load(os.path.join(root, p))
    plt.plot(data['e'], data['p_e'])
plt.yscale('log')
```

```python

for p in os.listdir(root):
    data = np.load(os.path.join(root, p))
    plt.scatter(data['N'], np.mean(data['edge_lengths']), c=np.log10(data['N_steps'] + 1))
plt.xscale('log')
plt.yscale('log')

N = np.logspace(1, 3)
plt.plot(N, 3 / np.sqrt(N), c='k', ls='--', label=r'$\langle L \rangle \sim \sqrt{N_v}$')

plt.legend()
plt.xlabel(r'$N_v$', fontsize=16)
plt.ylabel(r'$\langle L \rangle$', fontsize=16)
plt.tight_layout()
plt.savefig('figures/edge-length-scaling.png', dpi=600)
plt.show()
```
