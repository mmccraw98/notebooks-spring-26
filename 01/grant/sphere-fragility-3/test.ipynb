```python
import numpy as np
import matplotlib.pyplot as plt
from file_management import load_arrs
import os

path = '/home/mmccraw/dev/data/26-01-01/grant/sphere-fragilitiy/version-3/'
for cfg_type in os.listdir(path):
    data = load_arrs(os.path.join(path, cfg_type, 'phi-0.600000', 'traj', 'data.h5'))
    isf = data['isf']
    msd = data['msd']
    t = data['t']
    fig, ax = plt.subplots(1, 2, constrained_layout=True)
    plt.suptitle(cfg_type + ' ' + str(np.std(data['ke'] + data['pe']) / np.mean(data['ke'] + data['pe'])))
    ax[0].plot(t, msd)
    ax[1].plot(t, isf)
    for a in ax:
        a.set_xscale('log')
    ax[0].set_yscale('log')
    plt.show()
```
